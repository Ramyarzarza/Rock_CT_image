```python
from PIL import Image, ImageDraw
import random
import numpy as np

def generate_fracture_image(width=512, height=512, m_num_fractures=50, save_path="fracture_image.png"):
    # Create a grayscale image with a dark gray background
    # image = Image.new("L", (width, height), color=random.randint(20, 60))  # 'L' mode for grayscale, background = 50 (dark gray)
    # mask = Image.new("L", (width, height), color=0)
    image_n = str(random.randint(0, 117))
    image = Image.open('../../../Rock_CT_images/crack_segmentation_dataset/masks/' + image_n+'.jpg').convert("L")
    mask = Image.open('../../../Rock_CT_images/crack_segmentation_dataset/masks/' + image_n+'.jpg').convert("L")

    # Convert to NumPy array
    image = np.array(image)

    # Generate random numbers
    rand_zero = random.randint(20, 60)   # Replace 0 values with this
    rand_255 = random.randint(65, 150)  # Replace 255 values with this

    # Modify pixel values
    image[image <= 100] = rand_zero
    image[image >= 100] = rand_255

    # Convert back to PIL Image
    image = Image.fromarray(image)


    draw = ImageDraw.Draw(image)
    drawm = ImageDraw.Draw(mask)
    num_fractures = random.randint(0, m_num_fractures)

    # Optionally add some noise for realism
    for _ in range(random.randint(0, 200)):  # 1% noise
        x = random.randint(0, width - 1)
        y = random.randint(0, height - 1)
        noise_color = random.randint(55, 200)  # Slightly lighter than the background
        center = (x, y)  # (x, y)
        radius_x = random.randint(1, 9)
        radius_y = random.randint(1, 9)

        # Draw the filled circle
        draw.ellipse((center[0] - radius_x, center[1] - radius_y, 
                    center[0] + radius_x, center[1] + radius_y), fill=noise_color)
    # Generate random fractures
    for _ in range(num_fractures):
        # Random starting point for the fracture
        start_x = random.randint(0, width)
        start_y = random.randint(0, height)
        
        # Random length and direction
        length = random.randint(100, 300)
        angle = random.uniform(0, 2 * 3.14159)  # Random angle in radians
        
        # Calculate the end point
        end_x = int(start_x + length * random.uniform(0.2, 1.0) * random.choice([-1, 1]))
        end_y = int(start_y + length * random.uniform(0.2, 1.0) * random.choice([-1, 1]))

        # Random color for the fracture (lighter gray)
        color = random.randint(60, 200)  # Light gray color
        
        w = random.randint(1, 3)
        # Draw the fracture as a line
        draw.line((start_x, start_y, end_x, end_y), fill=color, width=w)
        drawm.line((start_x, start_y, end_x, end_y), fill=255, width=w)
    
    # Optionally add some noise for realism
    for _ in range(int(width * height * 0.01)):  # 1% noise
        x = random.randint(0, width - 1)
        y = random.randint(0, height - 1)
        noise_color = random.randint(80, 220)  # Slightly lighter than the background
        image.putpixel((x, y), noise_color)

    

    # Save the generated image
    image.save('../../../Rock_CT_images/synthetic/images/' + save_path)
    mask.save('../../../Rock_CT_images/synthetic/masks/'+ save_path)
    print(f"Image saved at {save_path}")

for i in range(5000):
    # Generate and save an example fracture image
    generate_fracture_image(448,448,40,str(i)+'.png')
```

```python
import os

def rename_images(directory):
    # Get all .png files in the directory
    images = [f for f in os.listdir(directory) if f.lower().endswith('.jpg')]
    
    # Sort images to maintain order
    images.sort()

    # Rename images with numbers
    for index, image in enumerate(images):
        old_path = os.path.join(directory, image)
        new_path = os.path.join(directory, f"{index}.jpg")
        os.rename(old_path, new_path)
        print(f"Renamed {image} -> {index}.jpg")

# Example usage
directory_path = "../../../Rock_CT_images/crack_segmentation_dataset/masks"  # Change this to your directory
rename_images(directory_path)
```
